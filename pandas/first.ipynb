```python
import pandas as pd
```

```python
# Create a sample DataFrame
data = {
    "name": ["Alice", "Bob", "Charlie"],
    "age": [25, 30, 35],
    "salary": [50000, 60000, 70000]
}

df = pd.DataFrame(data)
print(df)
```

```python
# it shows info about the DataFrame, including the number of non-null entries, data types, and memory usage.
df.info()
```

```python
df.describe()  # it provides summary statistics of the DataFrame, including count, mean, std, min, 25%, 50%, 75%, and max values for each numeric column.
```

```python
df.dtypes  # it shows the data types of each column in the DataFrame.
```

```python
# for Giving specific name to columns in dataframes
ronaldo = pd.DataFrame([11,22,33], columns=["Numbers"])
print(ronaldo)
```

```python
# it shows the column names of the DataFrame.
df.columns
```

```python
df.rename(columns={"name": "Name", "age": "Age", "salary": "Salary"}, inplace=True)  # it renames the columns of the DataFrame.
print(df)
```

```python
# how to save in CSV file
df.to_csv("data.csv", index=False)  # it saves the DataFrame to a CSV file named "data.csv" without the index column.
```

```python
read_data = pd.read_csv("data.csv")  # it reads the CSV file "data.csv" and returns a DataFrame.
print(read_data)
```

```python
# Selecting specific columns from the DataFrame
# read_data[["Name"]]
read_data[["Name","Salary"]]
```

```python
# Selecting specific row from dataframe using loc[]
# read_data.loc[read_data.Name == "Alice"]
# read_data.loc[read_data.Salary > 50000]
read_data.loc[(read_data.Age > 25) & (read_data.Salary > 60000)]
```

```python
# index value based way to select specific rows from the DataFrame using iloc[]
# read_data.iloc[1]  # it selects the second row of the DataFrame (index 1) and returns it as a Series.
read_data.iloc[1:3]  # it selects the second and third rows of the DataFrame (index 1 and 2) and returns them as a new DataFrame.
```

```python
# it will return a boolean Series indicating which rows have an "Age" greater than 30.
read_data["Age"] > 30
```

```python
read_data[read_data["Age"] > 30]  # it selects all rows from the DataFrame where the "Age" column is greater than 30 and returns them as a new DataFrame.
```

```python
# "Where" method is used to filter dataframe and it replace the other values with NaN
read_data.where(read_data["Age"] > 30)  
```

```python
# coustomize the the NaN values with a specific value using the "other" parameter.
read_data.where(read_data["Age"] > 30, other="Not Eligible")  
```

```python
# how to add columns in dataframe
read_data["Department"] = ["HR", "Finance", "IT"]
read_data
```

```python
read_data["Bonus"] = read_data["Salary"] * 0.1  # 10% bonus based on salary
read_data
```

```python
len(read_data)  # it returns the number of rows in the DataFrame.
```

```python
read_data.loc[len(read_data)] = ["Kashif", 28, 80000, "IT", 8000]  # it adds a new row to the DataFrame with the specified values for each column.
read_data
```

```python
read_data.loc[4, "Name"] = "Ronaldo"
print(read_data)  # it updates the value in the "Name" column of the row with index 4 to "Ronaldo".
```

```python
read_data.loc[read_data.Name == "Kashif", "Name"] = "Messi"
read_data
```

```python
read_data.drop(4, inplace=True)  # it removes the row with index 4 from the DataFrame.
print(read_data)
```

```python
# read_data.drop(5, inplace=True)  # it removes the row with index 4 from the DataFrame.
print(read_data)
```

```python
# it is used to remove a row
read_data.drop(read_data[read_data.Name == "Messi"].index, inplace=True)  # it removes the row(s) from the DataFrame where the "Name" column is equal to "Messi".
print(read_data)
```

```python
# how to delete a column from the DataFrame
read_data.drop("Bonus", axis=1, inplace=True)  # it removes the "Bonus" column from the DataFrame.
print(read_data)
```
